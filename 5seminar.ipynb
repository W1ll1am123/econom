```python
import numpy as np
import pandas as pd

from statsmodels.formula.api import ols

# Не показывать FutureWarnings
import warnings
warnings.simplefilter(action='ignore', category=Warning)
```

```python
# импорт данных
df = pd.read_csv('sleep75.csv')
df
```

```python
# спецификация модели через формулу
mod = ols(formula='sleep~1+totwrk+age+smsa', data=df)
# подгонка модели
res = mod.fit()
```

```python
# параметры подогнанной модели регрессии с округлением
res.params.round(2)
```

```python
# Создадим датафрейм с новыми данными регрессоров для прогноза
new_data = pd.DataFrame({'totwrk': [2150, 1950, 2240], 'age': [37, 26, 28], 'smsa': [1, 0, 1] })
new_data
```

```python
# Прогноз
res.predict(exog=new_data, transform=True).round(2)
#np.exp(res.predict(exog=new_data, transform=True))
```

```python
#3
# спецификация модели через формулу
df = pd.read_csv('wage2.csv')
df
mod = ols(formula='np.log(wage)~1+age+IQ+south+married+urban', data=df)
# подгонка модели
res = mod.fit()
res.params.round(3)
```

```python
# Создадим датафрейм с новыми данными регрессоров для прогноза
new_data = pd.DataFrame({'age': [36, 29, 25], 'IQ': [105, 123, 112],'south':[1,0,1], 'married': [1,1,0], 'urban':[1,0,1]})
new_data
```

```python
np.exp(res.predict(exog=new_data, transform=True))
```

```python
#5
# спецификация модели через формулу
df = pd.read_csv('Labour.csv')
df
mod = ols(formula='np.log(output)~1+np.log(capital)+np.log(labour)', data=df)
# подгонка модели
res = mod.fit()
res.params.round(3)
```

```python
#6
# спецификация модели через формулу
df = pd.read_csv('Labour.csv')
df
mod = ols(formula='np.log(output)~1+np.log(capital)+np.log(labour)+I(np.log(capital)**2)+I(np.log(labour)**2)', data=df)
# подгонка модели
res = mod.fit()
res.params.round(3)
```

```python
np.exp(res.predict(exog=new_data, transform=True))
```

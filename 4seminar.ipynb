```python
# подключение библиотек
import numpy as np
import pandas as pd

from statsmodels.formula.api import ols

# Не показывать Warnings
import warnings
warnings.simplefilter(action='ignore', category=Warning)
```

```python
# импорт данных из файла как DataFrame
df = pd.read_csv('sleep75.csv')
```

```python
# спецификация модели через формулу
mod = ols(formula='sleep~1+totwrk+age+smsa', data=df)
# подгонка модели
res = mod.fit()
```

```python
# параметры подогнанной модели регрессии
res.params
```

```python
# TSS
res.centered_tss
```

```python
# ESS
res.ess
```

```python
# RSS
res.ssr
```

```python
# R^2
res.rsquared
```

```python
# adj R^2
res.rsquared_adj
```

```python
# стандартная ошибка регрессии
res.scale**0.5
```

```python
# или стандартная ошибка регрессии
res.mse_resid**0.5
```

```python
# задаем индексы
ind = [1, 4, 6, 508, 700]
```

```python
# Фактические значения
df['sleep'].iloc[ind]
```

```python
# Предсказанные значения
res.fittedvalues[ind]
```

```python
# Остатки
res.resid[ind]
```

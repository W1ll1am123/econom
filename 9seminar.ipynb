```python
import pandas as pd
import numpy as np

from statsmodels.formula.api import ols
from statsmodels.iolib.summary2 import summary_col, summary_params # вывод результатов тестирования
from scipy.stats import t # t-распределение и критические значения
from scipy.stats import f # f-распределение и критические значения

# Не показывать Warnings
import warnings
warnings.simplefilter(action='ignore', category=Warning)
```

# 2.1

## 2.1.1

```python
# импорт данных
df = pd.read_csv('sleep75.csv')
```

```python
# спецификация модели
mod = ols(formula='sleep~totwrk+age+male+south+smsa+yngkid+marr+union', data=df)
# подгонка модели с ковариационной матрицей по умолчанию (неробастной)
res_ols = mod.fit()
# уровень значимости
sign_level = 0.1
# критическое значение t-распределения
t.ppf(q=1-sign_level/2, df=mod.df_resid)
```

```python
summary_params(res_ols, alpha=0.05).round(3)
```

```python
res_ols.pvalues.round(4)
```

```python
F_test = res_ols.f_test('smsa=yngkid=marr=union=0')
print(F_test)
```

```python
# уровень значимости
sign_level = 0.05
# Критическое значение F-распределения
f.ppf(q=1-sign_level, dfn=F_test.df_num, dfd=F_test.df_denom)
```

F < Fcr поэтому незначимо

## 2.2

```python
# спецификация модели
mod = ols(formula='sleep~totwrk+age+I(age**2)+male+south', data=df)
# подгонка модели с ковариационной матрицей по умолчанию (неробастной)
res_ols = mod.fit()
# уровень значимости
sign_level = 0.01
# критическое значение t-распределения
t.ppf(q=1-sign_level/2, df=mod.df_resid)
```

```python
summary_params(res_ols, alpha=0.01).round(3)
```

### 2.2.1

значимы totwrk, male, south т.к. у них |t| > tcr

```python
F_test = res_ols.f_test('age=I(age ** 2)=0')
print(F_test)
```

```python
# уровень значимости
sign_level = 0.01
# Критическое значение F-распределения
f.ppf(q=1-sign_level, dfn=F_test.df_num, dfd=F_test.df_denom)
```

Незначимы т.к. Fcr > F: 3>2.35

### 2.2.2

```python
res_hc = mod.fit(cov_type='HC3')
# альтернативная спецификация тестируемой гипотезы
F_test = res_hc.f_test('age=I(age ** 2)=0')
print(F_test)
```

```python
# Тестовая статистика и её P-значение
F_test.statistic, F_test.pvalue
```

```python
# уровень значимости
sign_level = 0.05
# Критическое значение F-распределения
f.isf(q=sign_level, dfn=F_test.df_num, dfd=F_test.df_denom)
```

Незначимо т.к. Fcr > F: 3 > 2.91
